```python
x = 1
while x < 101:
    if x % 3 == 0 and x % 5 == 0:
        print(x, "fizzbuzz")
    elif x % 3 == 0:
        print(x, "fizz")
    elif x % 5 == 0:
        print(x, "buzz")
    else:
        print(x)
    x = x + 1
```
